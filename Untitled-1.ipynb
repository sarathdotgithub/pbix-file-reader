```python
import zipfile
import os
import json

def explore_file(pbix_path):
    extract_to = "extracted_files"
    
    # 1. Extraction (Using the correct method: extractall)
    try:
        with zipfile.ZipFile(pbix_path, 'r') as zip_ref:
            zip_ref.extractall(extract_to)
            print(f"File extracted to: {extract_to}")
    except FileNotFoundError:
        print(f"Error: The file at {pbix_path} was not found.")
        return

    # 2. Path handling (Case-sensitive in some environments)
    layout_path = os.path.join(extract_to, "Report", "Layout")

    if os.path.exists(layout_path):
        with open(layout_path, 'r', encoding='utf-16-le') as f:
            layout_data = json.load(f)
            
            print("\n" + "="*40)
            print("KEY REPORT INSIGHTS")
            print("="*40)
            
            # Power BI stores pages in 'sections'
            for section in layout_data.get('sections', []):
                page_name = section.get('displayName')
                visuals = section.get('visualContainers', [])
                
                print(f"\n📄 PAGE: {page_name}")
                print(f"   - Total Visuals: {len(visuals)}")
                
                # Digging into individual visuals for insights
                visual_types = []
                data_fields = set()

                for vis in visuals:
                    config_str = vis.get('config')
                    if config_str:
                        config = json.loads(config_str)
                        
                        # Get the type of visual (e.g., pieChart, barChart)
                        vis_type = config.get('singleVisual', {}).get('visualType')
                        if vis_type:
                            visual_types.append(vis_type)
                        
                        # Extract data column names used in visuals
                        projections = config.get('singleVisual', {}).get('projections', {})
                        for key in projections:
                            for item in projections[key]:
                                query_ref = item.get('queryRef')
                                if query_ref:
                                    data_fields.add(query_ref.split('.')[-1])

                # Summarize findings for this page
                if visual_types:
                    unique_types = set(visual_types)
                    print(f"   - Chart Types: {', '.join(unique_types)}")
                
                if data_fields:
                    print(f"   - Fields Analyzed: {', '.join(list(data_fields)[:5])}...") # Top 5 fields

    else:
        print("Layout file not found. Check if the PBIX folder structure is standard.")

# Use the 'r' prefix to fix the path warning you saw earlier
explore_file(r"AdventureWorks PBIX Files\AdventureWorks Report_FINAL.pbix")
```

```python
import zipfile
import os
import json
import pandas as pd
import matplotlib.pyplot as plt

def visualize_pbix_structure(pbix_path):
    extract_to = "extracted_files"
    
    # 1. Extraction
    with zipfile.ZipFile(pbix_path, 'r') as zip_ref:
        zip_ref.extractall(extract_to)
    
    layout_path = os.path.join(extract_to, "Report", "Layout")
    
    visual_data = []

    if os.path.exists(layout_path):
        with open(layout_path, 'r', encoding='utf-16-le') as f:
            layout_data = json.load(f)
            
        for section in layout_data.get('sections', []):
            page_name = section.get('displayName')
            visuals = section.get('visualContainers', [])
            
            for vis in visuals:
                config_str = vis.get('config')
                if config_str:
                    config = json.loads(config_str)
                    vis_type = config.get('singleVisual', {}).get('visualType')
                    
                    if vis_type:
                        visual_data.append({
                            'Page': page_name,
                            'VisualType': vis_type
                        })

        # 2. Use Pandas to process the data
        df = pd.DataFrame(visual_data)
        
        if df.empty:
            print("No visuals found to visualize.")
            return

        # 3. Create a Summary Plot using Matplotlib
        plt.figure(figsize=(10, 6))
        df['VisualType'].value_counts().plot(kind='bar', color='skyblue', edgecolor='black')
        
        plt.title(f"Visual Composition: {os.path.basename(pbix_path)}", fontsize=14)
        plt.xlabel("Type of Power BI Visual", fontsize=12)
        plt.ylabel("Frequency", fontsize=12)
        plt.xticks(rotation=45)
        plt.grid(axis='y', linestyle='--', alpha=0.7)
        plt.tight_layout()
        
        # Display the chart
        plt.show()

        # 4. Show a summary table in the console
        print("\n--- Summary Table ---")
        print(df.groupby(['Page', 'VisualType']).size().unstack(fill_value=0))

    else:
        print("Layout file not found.")

# Execute the visualization
visualize_pbix_structure(r"AdventureWorks PBIX Files\AdventureWorks Report_FINAL.pbix")
```
